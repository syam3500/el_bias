# RRF Prediction Model for GC/MS Extractables — Surrogate level 20 µg/mL


## 0. Conception

**Inputs:** Physicochemical and molecular descriptors computed from the compound SMILES (RDKit descriptors) together with experimental properties sourced from the FDA CLAP list and Yun et al. (2022).

**Outcome:** RRF measured at 20 µg/mL (continuous, positive variable).

### Success criteria

Models are compared on 10-fold cross-validated R² and RMSE. The best model is the one that maximises R² and minimises test-set RMSE on the hold-out set (30 %). Predictions with R² > 0.7 on the test set are considered acceptable for correcting quantification bias in a screening context.

## 4. Data Preprocessing

Preprocessing converts raw chemical identifiers into a structured feature matrix suitable for machine learning. The main steps are:

1. Retrieve the canonical SMILES string for each compound via PubChemPy.
2. Compute the full set of 2D RDKit molecular descriptors (~200 features) from each SMILES.
3. Expand the descriptor dictionary into individual columns and join with the original dataset.


### 4.1 CAS → SMILES conversion

The CAS Registry Number is used as a query key to retrieve the canonical SMILES representation of each compound from the PubChem database via the `pubchempy` API. SMILES (Simplified Molecular Input Line Entry System) encodes the molecular graph and is the required input format for RDKit descriptor computation.

```python
def get_smiles_from_cas(cas):
    """Utility to fetch SMILES via PubChem"""
    try:
        results = pcp.get_compounds(cas, 'name')
        if results:
            return results[0].smiles
    except Exception as e:
        return None
    return None

df['SMILES'] = df['CAS #'].apply(get_smiles_from_cas)
```

> **Manual correction**: The compound with CAS 68515-75-3 (*dioctyl sebacate*, a common plasticiser) could not be resolved automatically via PubChem. Its SMILES was manually verified and inserted to avoid losing this compound from the training set.

```python
target_cas = '68515-75-3'
manual_smiles = 'CCCCCCCCCOC(=O)CCCCC(=O)OCCCCCCC'

df.loc[df['CAS #'] == target_cas, 'SMILES'] = manual_smiles
```

### 4.3 RDKit molecular descriptor computation

For each canonical SMILES, the full suite of 2D RDKit descriptors is computed using `MolecularDescriptorCalculator`. These descriptors encode diverse structural and physicochemical properties of the molecule — including lipophilicity (`MolLogP`), molecular weight (`MolWt`), topological polar surface area (`TPSA`), and a range of topological and electronic indices — without requiring 3D coordinates. Any molecule for which SMILES parsing fails returns `None` and is excluded from the descriptor matrix.

```python
def standardize_and_describe(smiles):
    if not smiles: return None
    
    mol = Chem.MolFromSmiles(smiles)
    if mol is None: return None
    
    try:
        desc_names = [d[0] for d in Descriptors._descList]
        calculator = MoleculeDescriptors.MolecularDescriptorCalculator(desc_names)
        
        vals = calculator.CalcDescriptors(mol)
        return dict(zip(desc_names, vals))
        
    except:
        return None
    
results = df['SMILES'].apply(standardize_and_describe)
results[0]
```

### 4.2 Joining molecular descriptors to the dataset

The `standardize_and_describe` function returns a dictionary of ~200 RDKit descriptors for each molecule. Here, each dictionary is expanded into individual columns and joined to the original dataframe. After this step, `df_final` contains both the experimental properties from the CLAP list/Yun et al. and the computationally derived structural descriptors (e.g., `MolLogP`, `TPSA`, `MolWt`, topological indices).

```python
df_final = df.join(df['SMILES'].apply(standardize_and_describe).apply(pd.Series))
```

## 6. Feature Engineering & Train/Test Split

### 6.1 Handling missing values — Multiple Imputation

Missing values are handled via **Multiple Imputation by Chained Equations (MICE)**, implemented as `IterativeImputer` in scikit-learn. MICE iteratively models each feature with missing data as a function of the other features, producing unbiased imputations under the Missing At Random (MAR) assumption. Ten iterations are used (`max_iter=10`) to ensure convergence.

### 6.2 Feature selection strategy

Given the large number of RDKit descriptors (~200), a preliminary feature selection step is applied to reduce dimensionality and mitigate the curse of dimensionality. A preliminary Random Forest (100 trees) is fitted on the training set and the **top 20 features by Gini importance** are retained. This subset (`X_train_top`) is used for the tree-based models (Random Forest, LightGBM), while the full scaled feature matrix (`X_train_sel`) is used for Elastic Net and SVM, which benefit from the regularisation rather than explicit feature selection.

*Note: A version using all available descriptors without prior selection also exists in the archive for comparison.*

```python
missing_data = df_final.isnull().sum()
missing_data = missing_data[missing_data > 0].sort_values(ascending=False)

print(missing_data)
```

Missing values exist for `pKA` (27),  `Refractive Index`(3), `BP`(1).
Values for RI and BP will be imputed using Multiple Imputation. pKA will not be used in the models.

```python
X = df_final.drop(['5 µg/mL', 'IUPAC name', 'CAS #', 'Empirical Formula', '10 µg/mL', '20 µg/mL', 'Code', 'pKa', 'SMILES'] , axis=1) # Features without the 20 µg/mL RRF
y = df_final['20 µg/mL'] # outcome

# Multiple Imputation using IterativeImputer
imputer = IterativeImputer(max_iter=10, random_state=42)
scaler = StandardScaler()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)

# Impute and Scale (for Elastic Net and SVM)
X_train_imp = imputer.fit_transform(X_train)
X_test_imp = imputer.transform(X_test)

X_train_scaled = scaler.fit_transform(X_train_imp)
X_test_scaled = scaler.transform(X_test_imp)

# Keep as DataFrames for feature selection reference
feature_names = X.columns
X_train_df = pd.DataFrame(X_train_scaled, columns=feature_names)
X_test_df = pd.DataFrame(X_test_scaled, columns=feature_names)


X_train_sel = X_train_df
X_test_sel = X_test_df

temp_rf = RandomForestRegressor(n_estimators=100, random_state=42)
temp_rf.fit(X_train_df, y_train)
importances = pd.Series(temp_rf.feature_importances_, index=feature_names)
top_20_features = importances.sort_values(ascending=False).head(20).index

X_train_top = X_train_df[top_20_features]
X_test_top = X_test_df[top_20_features]
```

## 7. Model Selection

Given a continuous, bounded outcome (RRF > 0), we evaluate four complementary regression approaches spanning parametric and non-parametric families:

| Model | Type | Notes |
|---|---|---|
| **Elastic Net** | Parametric, linear | Combined L1/L2 regularisation; interpretable coefficients; requires scaled inputs |
| **Random Forest** | Non-parametric, ensemble | Robust to outliers; handles non-linear interactions; feature importance available |

| **Support Vector Machine (RBF)** | Non-parametric, kernel | Effective in high-dimensional spaces; sensitive to feature scaling |

## 9. Baseline Model Evaluation (10-fold Cross-Validation)

Before hyperparameter tuning, each model is evaluated via 10-fold stratified cross-validation on the training set (`cross_val_predict`). This provides unbiased estimates of R², MAE, and RMSE reflecting the generalisation performance of each model architecture with default settings.

```python
kf = KFold(n_splits=10, shuffle=True, random_state=42)

def evaluate_model_cv(model, X, y, model_name="Model"):
    cv_predictions = cross_val_predict(model, X, y, cv=kf, n_jobs=-1)
    
    cv_r2 = r2_score(y, cv_predictions)
    cv_mae = mean_absolute_error(y, cv_predictions)
    cv_rmse = np.sqrt(mean_squared_error(y, cv_predictions))
    
    print(f"--- {model_name} (k-Fold CV) ---")
    print(f"R²:   {cv_r2:.4f}")
    print(f"MAE:  {cv_mae:.4f}")
    print(f"RMSE: {cv_rmse:.4f}\n")
    
    return cv_predictions

cv_results = {}
for name, model in models.items():
    if name in ['Elastic Net', 'SVM']:
        cv_results[name] = evaluate_model_cv(model, X_train_sel, y_train, name)
    else:
        cv_results[name] = evaluate_model_cv(model, X_train_top, y_train, name)
```

## 10. Hyperparameter Optimisation (Bayesian Search with Optuna)

The Elastic Net was already implicitly optimised during the feature engineering step (regularisation path). For Random Forest, LightGBM, and SVM, we perform **Bayesian optimisation** using the `optuna` framework, which models the hyperparameter–objective relationship as a Gaussian process and proposes parameter combinations that maximise the expected improvement over previous trials. The objective function is the 5-fold cross-validated R² on the training set.

- **SVM**: 30 trials optimising `C`, `epsilon`, and `gamma`.
- **Random Forest**: 30 trials optimising `n_estimators`, `max_depth`, and `min_samples_split`.
- **LightGBM**: 50 trials optimising regularisation, leaf count, `feature_fraction`, and `bagging` parameters.

```python
def objective_svm(trial):
    c = trial.suggest_float('C', 1e-3, 1e2, log=True)
    epsilon = trial.suggest_float('epsilon', 1e-3, 0.5, log=True)
    gamma = trial.suggest_categorical('gamma', ['scale', 'auto'])
    
    model = SVR(C=c, epsilon=epsilon, gamma=gamma)
    
    score = cross_val_score(model, X_train_sel, y_train, cv=5, scoring='r2').mean()
    return score

study_svm = optuna.create_study(direction='maximize')
study_svm.optimize(objective_svm, n_trials=30)

def objective_rf(trial):
    n_estimators = trial.suggest_int('n_estimators', 50, 1000)
    max_depth = trial.suggest_int('max_depth', 2, 32, log=True)
    min_samples_split = trial.suggest_int('min_samples_split', 2, 20)
    
    model = RandomForestRegressor(
        n_estimators=n_estimators, 
        max_depth=max_depth, 
        min_samples_split=min_samples_split,
        random_state=42
    )
    
    score = cross_val_score(model, X_train_top, y_train, cv=5, scoring='r2').mean()
    return score

study_rf = optuna.create_study(direction='maximize')
study_rf.optimize(objective_rf, n_trials=30)

def objective_lgbm(trial):
    param = {
        'objective': 'regression',
        'metric': 'rmse',
        'verbosity': -1,
        'boosting_type': 'gbdt',
        'lambda_l1': trial.suggest_float('lambda_l1', 1e-8, 10.0, log=True),
        'lambda_l2': trial.suggest_float('lambda_l2', 1e-8, 10.0, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 2, 256),
        'feature_fraction': trial.suggest_float('feature_fraction', 0.4, 1.0),
        'bagging_fraction': trial.suggest_float('bagging_fraction', 0.4, 1.0),
        'bagging_freq': trial.suggest_int('bagging_freq', 1, 7),
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 100),
    }

    model = LGBMRegressor(**param)
    score = cross_val_score(model, X_train_top, y_train, cv=5, scoring='r2').mean()
    return score

study_lgbm = optuna.create_study(direction='maximize')
study_lgbm.optimize(objective_lgbm, n_trials=50)
```

```python
def plot_model_performance(y_true, y_pred, model_name):
    """
    Generates Predicted vs Actual and Residuals plots
    """
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    
    # Plot 1: Predicted vs Actual
    
    axes[0].scatter(y_true, y_pred, alpha=0.6, color='blue', edgecolor='k')
    
    # Perfect prediction line
    min_val = min(y_true.min(), y_pred.min())
    max_val = max(y_true.max(), y_pred.max())
    axes[0].plot([min_val, max_val], [min_val, max_val], 'r--', lw=2)
    
    axes[0].set_title(f'{model_name}: Predicted vs Actual')
    axes[0].set_xlabel('Actual RRF')
    axes[0].set_ylabel('Predicted RRF')
    axes[0].grid(True)

    # Plot 2: Residuals
    residuals = y_true - y_pred
    axes[1].scatter(y_pred, residuals, alpha=0.6, color='purple', edgecolor='k')
    axes[1].axhline(y=0, color='r', linestyle='--', lw=2)
    
    axes[1].set_title(f'{model_name}: Residual Plot')
    axes[1].set_xlabel('Predicted RRF')
    axes[1].set_ylabel('Residuals (Actual - Predicted)')
    axes[1].grid(True)
    
    plt.tight_layout()
    plt.show()
```

- **Predicted vs Actual**: points should lie close to the 45° identity line.
- **Residual plot**: residuals should be randomly scattered around zero with no systematic pattern, which would indicate model misspecification.

## 12. Model Persistence & Reproducibility

The best-performing pipeline is serialised for subsequent inference. A `sklearn.pipeline.Pipeline` object is constructed to bundle preprocessing (imputation → scaling) and the final Random Forest model. Serialising the complete pipeline ensures that new predictions are pre-processed identically to the training data, avoiding train/test leakage.

Exported artefacts:

- `rf_rrf_model_final_20_20ug.joblib` — complete preprocessing + RF pipeline.
- `features_20_20ug.joblib` — ordered list of the 20 selected feature names required at inference time.
- `final_model_data_20ug.csv` — final processed dataset (including descriptors) for traceability.


```python
import joblib
from sklearn.pipeline import Pipeline
from sklearn.impute import IterativeImputer
from sklearn.preprocessing import StandardScaler

importances = pd.Series(best_rf.feature_importances_, index=feature_names)
top_20_features_list = list(top_20_features) # Ensure it's a list

# Build pipeline to handle Imputation and Scaling dynamically
final_pipeline = Pipeline([
    ('imputer', IterativeImputer(max_iter=10, random_state=42)),
    ('scaler', StandardScaler()),
    ('rf', RandomForestRegressor(**study_rf.best_params, random_state=42, n_jobs=-1))
])

# Fit on unscaled, raw top 20 features
X_train_raw_top = X_train[top_20_features_list]
final_pipeline.fit(X_train_raw_top, y_train)

joblib.dump(final_pipeline, 'rf_rrf_model_final_20_20ug.joblib')
joblib.dump(top_20_features_list, 'features_20_20ug.joblib')

df_final.to_csv("final_model_data_20ug.csv", index=False)
```

## 13. Summary

This notebook trained and compared four regression models to predict the GC/MS Relative Response Factor at 20 µg/mL. The workflow covered:

1. **Data acquisition**: 20 µg/mL RRF values from the FDA CLAP list and Yun et al.
2. **Descriptor computation**: ~200 RDKit 2D molecular descriptors computed from canonical SMILES retrieved via PubChem.
3. **Preprocessing**: multiple imputation (MICE) for `RI` and `BP`; standard scaling for Elastic Net and SVM.
4. **Feature selection**: top 20 features by Gini importance from a preliminary Random Forest.
5. **Modelling**: Elastic Net, Random Forest, LightGBM, and SVM trained and compared by 10-fold CV R² and RMSE.
6. **Hyperparameter optimisation**: Bayesian search (Optuna) for RF, LGBM, and SVM.
7. **Model export**: complete sklearn Pipeline persisted to `rf_rrf_model_final_20_20ug.joblib` for use in concentration correction of unknown extractables.

The predicted RRFs are consumed downstream by the Quarto (`.qmd`) reporting pipeline.
